## Examine state codes

```python
# a. Non-NA values in the State column
df.loc[df["State"].notna(), "State"]
```

b. Challenge: Some entries contain multiple states in a single cell separated by semicoloons ("CA;NV"). Calling .unique() directly would treat "CA:NV" and "CA" as distinct values, so you'd get duplicated states in compound strings rather than a clean list of individual states.

## Exploratory Wrangling

```python
df.loc[df['Conflict Present'] == 'Y', 'State'].str.split(';').explode()
```

```python
# Select state column and rows with water conflict
# Then, split on semicolon delimiter
# Then, explode lists into individual rows
# Then, display unique values

df.loc[df['Conflict Present'] == 'Y', 'State'].str.split(';').explode().unique()
```

## 9. Find unique state codes

```python
states_unique = (
    df.loc[df["Conflict Present"] == "Y", "State"]
    .str.split(";")
    .explode()
    .str.strip()
    .unique()
)
states_unique
```

## 10. Visualize Data Exploration and Discussion

```python
states_con_val = (
    df.loc[df["Conflict Present"] == "Y", "State"]
    .str.split(";")
    .explode()
    .str.strip()
    .value_counts(dropna=False)
)
states_con_val
```

```python
states_con_val.plot(kind = 'bar',
                    x = 'State',
                   title = 'Number of conflicts by state',
                   xlabel = 'State',
                   ylabel = 'Number of conflicts')
```
